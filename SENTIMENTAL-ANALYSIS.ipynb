```python
from transformers import pipeline
```

```python
# Load the classification pipeline with the specified model
model = pipeline("text-classification", model="tabularisai/multilingual-sentiment-analysis")
```

```python
# Classify a new sentence
review = """
 I recently visited  for a family dinner and was thoroughly impressed. The margarita pizza had a beautifully charred, thin crust, and the fresh basil really made the flavors pop. Our waiter was knowledgeable and prompt, even during the Friday night rush. The lively, modern atmosphere made for a fun evening, and I’ll definitely be back soon!
"""


# Print the result
result = model(review)
print(result)
```

```python
from transformers import pipeline
classifier = pipeline('zero-shot-classification', model='roberta-large-mnli')
```

```python
review = """
 I recently visited [Restaurant Name] for a family dinner and was thoroughly impressed. The margarita pizza had a beautifully charred, thin crust, and the fresh basil really made the flavors pop. Our waiter was knowledgeable and prompt, even during the Friday night rush. The lively, modern atmosphere made for a fun evening, and I’ll definitely be back soon!
"""

candidate_labels = ['travel', 'positive', 'very positive', 'negative']

classifier(review, candidate_labels)['labels'][0]
```
